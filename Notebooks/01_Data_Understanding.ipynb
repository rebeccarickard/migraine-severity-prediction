### Variable Descriptions

```python
import numpy as np

# Variables grouped according to their role in the study

target_variables = {
    "Severity",
    "Intensity (Mild vs Moderate/Severe)",
    "Index (VAS Style Score)"
}

identifier_variables = {
    "No.",
    "Registration No.",
    "Unique ID"
}

time_variables = {
    "Date",
    "Start Time",
    "Start Time.1",
    "End Time",
    "End Time.1",
    "Duration",
    "Duration (min)"
}

physiological_variables = {
    "Preventive Medication Use",
    "Acute Medication Use",
    "Internal Factor: Stress",
    "Internal Factor: Oversleeping",
    "Internal Factor: Sleep Deprivation",
    "Internal Factor: Exercise",
    "Internal Factor: Lack of Exercise",
    "Internal Factor: Physical Fatigue",
    "Internal Factor: Menstrual Cycle: Menstruation",
    "Internal Factor: Menstrual Cycle: Ovulation",
    "Internal Factor: Excessive Emotional Change",
    "Other Factor: Irregular Meals (fasting, etc.)",
    "Exercise (Y/N)",
    "Vigorous Exercise",
    "Moderate Exercise",
    "Total Exercise",
    "Vigorous Exercise (min)",
    "Moderate Exercise (min)"
}

environmental_variables = {
    "External Factor: Weather/Temperature Change",
    "External Factor: Excessive Sunlight",
    "External Factor: Noise",
    "External Factor: Inappropriate Lighting",
    "External Factor: Specific Smell (cosmetics, perfume, etc.)"
}

outcome_or_post_migraine_variables = {
    "Headache-free Day",
    "Headache Ongoing (Y/N)",
    "Impacts Daily Life",
    "Impacts Daily Life.1",
    "Missed Work/School",
    "Productivity Reduced by Half or More",
    "Total Impact on Work or School",
    "Unable to Do Housework at All",
    "Housework Efficiency Reduced by Half or More",
    "Total Impact at Home",
    "Social/Leisuire: Planned but Unable to Participate",
    "No Effect from Acute Medication",
    "Headache Reduced by Half or More with Acute Medication",
    "Headache Resolved with Acute Medication"
}

def assign_variable_category(column_name):
    if column_name in target_variables:
        return "Target Variable"
    if column_name in identifier_variables:
        return "Identifier Variable"
    if column_name in time_variables:
        return "Date/Time"
    if column_name in physiological_variables:
        return "Physiological"
    if column_name in environmental_variables:
        return "Environmental"
    if column_name in outcome_or_post_migraine_variables:
        return "Outcome/Post-Migraine"
    if column_name.startswith("Pain"):
        return "Headache characteristic"
    if "Headache Type" in column_name or "Headache" in column_name:
        return "Headache characteristic"
    if "Accompanying Symptom" in column_name:
        return "Associated symptom"
    if column_name.startswith("Relief Factor"):
        return "Relief factor"
    if column_name.startswith("Internal Factor"):
        return "Physiological"
    if column_name.startswith("External Factor"):
        return "Environmental"
    if column_name.startswith("Other Factor"):
        return "Other trigger"
    if "Medication" in column_name or column_name in {
        "Tylenol (Acetaminophen)",
        "Combination Drug",
        "Triptan",
        "Other Medication Name"
    }:
        return "Medication"
    return "Other Variable"

variable_descriptions = {
    "No.": "Sequential record number.",
    "Registration No.": "Participant registration number.",
    "Unique ID": "Unique participant identifier.",
    "Date": "Date associated with the diary record.",
    "Start Time": "Recorded starting time of the headache episode.",
    "End Time": "Recorded ending time of the headache episode.",
    "Duration": "Recorded duration of the headache episode.",
    "Duration (min)": "Duration of the headache episode measured in minutes.",

    "Severity": (
        "Reported migraine or headache severity. This is the primary "
        "outcome variable for severity prediction."
    ),
    "Intensity (Mild vs Moderate/Severe)": (
        "Binary grouping that distinguishes mild headaches from "
        "moderate or severe headaches."
    ),
    "Index (VAS Style Score)": (
        "Numeric pain-intensity measure based on a visual analog-style scale."
    ),

    "Preventive Medication Use": (
        "Indicates whether preventive migraine medication was used."
    ),
    "Acute Medication Use": (
        "Indicates whether acute medication was used in response to a headache."
    ),
    "Internal Factor: Stress": (
        "Indicates whether stress was reported as a trigger associated "
        "with the headache episode."
    ),
    "Internal Factor: Oversleeping": (
        "Indicates whether oversleeping was reported as a headache trigger."
    ),
    "Internal Factor: Sleep Deprivation": (
        "Indicates whether insufficient sleep was reported as a headache trigger."
    ),
    "Internal Factor: Exercise": (
        "Indicates whether exercise was reported as a headache trigger."
    ),
    "Internal Factor: Lack of Exercise": (
        "Indicates whether lack of exercise was reported as a headache trigger."
    ),
    "Internal Factor: Physical Fatigue": (
        "Indicates whether physical fatigue was reported as a headache trigger."
    ),
    "Internal Factor: Menstrual Cycle: Menstruation": (
        "Indicates whether menstruation was reported as a headache trigger."
    ),
    "Internal Factor: Menstrual Cycle: Ovulation": (
        "Indicates whether ovulation was reported as a headache trigger."
    ),
    "Internal Factor: Excessive Emotional Change": (
        "Indicates whether a substantial emotional change was reported "
        "as a headache trigger."
    ),
    "Other Factor: Irregular Meals (fasting, etc.)": (
        "Indicates whether fasting, meal skipping, or irregular meal timing "
        "was reported as a headache trigger."
    ),

    "External Factor: Weather/Temperature Change": (
        "Indicates whether a change in weather or temperature was reported "
        "as a headache trigger."
    ),
    "External Factor: Excessive Sunlight": (
        "Indicates whether excessive sunlight was reported as a headache trigger."
    ),
    "External Factor: Noise": (
        "Indicates whether environmental noise was reported as a headache trigger."
    ),
    "External Factor: Inappropriate Lighting": (
        "Indicates whether lighting conditions were reported as a headache trigger."
    ),
    "External Factor: Specific Smell (cosmetics, perfume, etc.)": (
        "Indicates whether exposure to a particular smell was reported "
        "as a headache trigger."
    ),

    "Exercise (Y/N)": (
        "Indicates whether the participant exercised during the relevant period."
    ),
    "Vigorous Exercise": "Recorded amount or occurrence of vigorous exercise.",
    "Moderate Exercise": "Recorded amount or occurrence of moderate exercise.",
    "Total Exercise": "Combined measure of exercise activity.",
    "Vigorous Exercise (min)": (
        "Amount of vigorous exercise measured in minutes."
    ),
    "Moderate Exercise (min)": (
        "Amount of moderate exercise measured in minutes."
    ),

    "Headache-free Day": (
        "Indicates whether the diary date was classified as headache-free."
    ),
    "Headache Ongoing (Y/N)": (
        "Indicates whether a headache was ongoing at the time of the record."
    )
}

data_dictionary = pd.DataFrame({
    "Variable": diary.columns,
    "Current Type": diary.dtypes.astype(str).values,
    "Category": [
        assign_variable_category(column)
        for column in diary.columns
    ],
    "Description": [
        variable_descriptions.get(
            column,
            "Description to be reviewed and added."
        )
        for column in diary.columns
    ],
    "Non-Missing Values": diary.notna().sum().values,
    "Missing Values": diary.isna().sum().values,
    "Percent Missing": (
        diary.isna().mean().mul(100).round(2).values
    ),
    "Unique Values": diary.nunique(dropna=True).values
})

data_dictionary.head(20)
```

### Initial Observations

- The diary dataset consists of approximately 4500 daily observations (4579 exact) from 62 migraine patients. Close to 1100 headache/migraine events (1099 exact) were recorded.
- The target variable consists of three severity categories: mild, moderate, and severe. Multiple physiological and environmental predictors were used, such as stress, sleep, menstrual cycle, exercise, dietary habits, medication use, excessive light, excessive smells, and loud sound. Self reported weather as a trigger was also used as a predictor, but actual weather conditions were not included.
- Preliminary examination reveals:
    - There are missing values related to the presence/absence of a headache/migraine (~1099 present/~3480 absent)
    - Any missing values that do not fall into this pattern should the be the highest concern when handling missing values
    - 2 start times (Start Time and Start Time.1) and 2 end times (End Time and End Time.1). Time.1 is from diary dataset and is the hour-time the migraine began and ended. The other Start/End Time is date/time format
    - Some variables in Date/Time category are not the same type (float, object, int). Address this during preprocessing
    - Target variables are unbalanced with severe having a much smaller representation than mild and moderate:
        - Mild: 547
        - Moderate: 402
        - Severe: 117
        - This distribution is to be expected. **Not sure how to address this throughout the project**
